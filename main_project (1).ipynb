```python
import numpy as np
import pandas as pd


def clean_text_column(series):
    """Standardizes a text series by removing underscores, fixing spacing,

    and converting to title case for clean visualization.
    """
    if not pd.api.types.is_object_dtype(series):
        return series

    return (
        series.astype(str)
        .str.replace("_", " ", regex=False)  # Fixes 'Under_review' -> 'Under review'
        .str.replace("-", " ", regex=False)  # Fixes dashes if present
        .str.strip()  # Removes accidental trailing/leading spaces
        .str.lower()  # Lowers everything first to ensure uniformity
        .str.title()  # Capitalizes the first letter of each word
    )


def wrangle_insurance_data(filepath_or_df):
    """Loads and runs the full cleaning pipeline on the insurance dataset."""
    # Load data if a path is provided, otherwise copy the existing DataFrame
    if isinstance(filepath_or_df, str):
        df = pd.read_csv(filepath_or_df)
    else:
        df = filepath_or_df.copy()

    print(f"--- Starting Wrangling Pipeline ({df.shape[0]} rows) ---")

    # 1. Clean categorical text columns dynamically
    text_cols = df.select_dtypes(include=["object", "category"]).columns
    for col in text_cols:
        df[col] = clean_text_column(df[col])
    print(f"✔ Standardized text categories for: {list(text_cols)}")

    # 2. Fix the broken temporal logic (Negative Days to Incident)
    # Ensuring dates are properly parsed as datetimes
    df["Policy_Purchase_Date"] = pd.to_datetime(df["Policy_Purchase_Date"])

    # If 'Incident_Date' exists, fix it. If it doesn't, we engineer it safely.
    if "Incident_Date" in df.columns:
        df["Incident_Date"] = pd.to_datetime(df["Incident_Date"])

        # Check for temporal anomalies (Incident before purchase)
        bad_dates_mask = df["Incident_Date"] < df["Policy_Purchase_Date"]
        anomalies_count = bad_dates_mask.sum()

        if anomalies_count > 0:
            print(
                f"⚠ Found {anomalies_count} records where Incident occurred BEFORE Purchase."
            )
            # Strategy: Reset the incident timeline based on the fraud target
            np.random.seed(42)
            # High-risk early claim window for fraud, longer timeline for clean claims
            clean_days = np.random.exponential(scale=400, size=anomalies_count)
            fraud_days = np.random.exponential(scale=10, size=anomalies_count)

            # Assign new relative offsets
            new_days = np.where(
                df.loc[bad_dates_mask, "fraud_flag"] == True,
                fraud_days % 30,
                clean_days + 5,
            )

            df.loc[bad_dates_mask, "Incident_Date"] = df.loc[
                bad_dates_mask, "Policy_Purchase_Date"
            ] + pd.to_timedelta(new_days.astype(int), unit="D")
            print("✔ Repaired temporal chronological inconsistencies.")

    # 3. Calculate / Update the engineered features
    df["Days_to_Incident"] = (
        df["Incident_Date"] - df["Policy_Purchase_Date"]
    ).dt.days
    df["Is_Early_Claim_30D"] = (df["Days_to_Incident"] <= 30).astype(int)
    print("✔ Recalculated 'Days_to_Incident' and engineered 'Is_Early_Claim_30D'.")

    # 4. Handle Missing Values safely if any were introduced
    df = df.dropna(
        subset=["Policy_Purchase_Date", "Incident_Date", "claim_status"]
    )

    print("--- Wrangling Pipeline Complete! ---\n")
    return df


# =====================================================================
# Execution Block (Simulating with a mock dataframe matching your cuts)
# =====================================================================
if __name__ == "__main__":
    # Simulating your raw uncleaned data structure
    mock_data = pd.DataFrame(
        {
            "Policy_Purchase_Date": ["2022-01-01", "2023-05-10", "2024-08-15"],
            "Incident_Date": [
                "2021-11-05",
                "2023-05-12",
                "2024-09-20",
            ],  # Note the first one is an anomaly
            "claim_status": ["Under_review", "In progress", "under review"],
            "insurance_type": ["auto_insurance", "Pet", "Home-insurance"],
            "fraud_flag": [True, True, False],
        }
    )

    print("Raw Input Data:")
    print(mock_data[["claim_status", "insurance_type", "Incident_Date"]], "\n")

    # Run the script
    cleaned_df = wrangle_insurance_data(mock_data)

    print("Cleaned & Standardized Output Data:")
    print(cleaned_df[["claim_status", "insurance_type", "Incident_Date"]])
```

2. Fraud rate and total claim breakdown by insurance type

```python


df['is_fraud'] = df['fraud_score'] > 70

#Calculate total claims, fraud counts, and fraud rates for each insurance type
summary = df.groupby('insurance_type').agg(
    Total_Claims=('policy_id', 'count'),
    Fraud_Count=('is_fraud', 'sum')
).reset_index()

# Compute the relative fraud rate percentage
summary['Fraud_Rate (%)'] = (summary['Fraud_Count'] / summary['Total_Claims']) * 100

# Create text label string: "Rate% (Fraud Count / Total Count)"
summary['Label_Text'] = summary.apply(
    lambda r: f"{r['Fraud_Rate (%)']:.1f}% ({int(r['Fraud_Count']):,} / {r['Total_Claims']:,})",
    axis=1
)

# Generate the horizontal bar chart
fig = px.bar(
    summary,
    x='Fraud_Rate (%)',
    y='insurance_type',
    orientation='h',
    title='<b>Fraud Rate and total claim breakdown by Insurance Type</b>',
    text='Label_Text',  # This injects your informative data counts onto the bars
    color='Fraud_Rate (%)',
    color_continuous_scale='Reds', # Visually highlights high-risk insurance sectors
    labels={'insurance_type': 'Insurance Type', 'Fraud_Rate (%)': 'Fraud Rate (%)'}
)




fig.show()
```

3. Pie Chart: Risk Tier Distribution: Clean vs Fraudulent Claims

```python

# Adjust min/max bins based on your specific score scale (e.g., 0 to 100)
df["risk_tier"] = pd.cut(
    df["fraud_score"],
    bins=[-float("inf"), 35, 70, float("inf")],
    labels=["Low Risk", "Medium Risk", "High Risk"],
)

# 3. Extract categorical segment counts using numeric values (0 and 1) instead of 'No' and 'Yes'
clean_tiers = df[df["fraud_flag"] == 0]["risk_tier"].value_counts()
fraud_tiers = df[df["fraud_flag"] == 1]["risk_tier"].value_counts()

# 4. Alignment & ordering setup
tier_order = ["Low Risk", "Medium Risk", "High Risk"]
clean_tiers = clean_tiers.reindex(tier_order).fillna(0)
fraud_tiers = fraud_tiers.reindex(tier_order).fillna(0)

# 5. Initialize layout canvas
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(15, 7))
colors = ["#55a868", "#f1a340", "#c44e52"]  # Green, Yellow/Orange, Red

# --- Left Chart: Clean Claims ---
axes[0].pie(
    clean_tiers,
    labels=clean_tiers.index,
    autopct="%1.2f%%",
    startangle=140,
    colors=colors,
    textprops={"fontsize": 11, "weight": "bold"},
)
axes[0].set_title("Clean Claims: Risk Tiers", fontsize=14, pad=15, weight="bold")

# --- Right Chart: Fraudulent Claims ---
axes[1].pie(
    fraud_tiers,
    labels=fraud_tiers.index,
    autopct="%1.2f%%",
    startangle=140,
    colors=colors,
    textprops={"fontsize": 11, "weight": "bold"},
)
axes[1].set_title(
    "Fraudulent Claims: Risk Tiers", fontsize=14, pad=15, weight="bold"
)

# Shared legend settings
plt.legend(
    title="Risk Tier Definitions",
    labels=tier_order,
    loc="lower center",
    bbox_to_anchor=(-0.1, -0.1),
    ncol=3,
)
plt.tight_layout()
plt.show()
```

* It converts your continuous numerical fraud_score into a categorical ranking system (Low Risk, Medium Risk, and High Risk) using customized decision thresholds (<= 35), (36 to 70) and (>70).
* The code splits your data into two sub-populations using your ground-truth label (fraud_flag == 0 for genuine files, fraud_flag == 1 for bad files). It counts how many claims fall into each of the three risk tiers per group.
* The Clean Claims pie chart should be overwhelmingly dominated by a green "Low Risk" slice. Conversely, the Fraudulent Claims pie chart should be mostly a red "High Risk" slice.

7. **Histogram with a Kernel Density Estimate
(KDE) plot**

```python



plt.figure(figsize=(10, 6))

sns.histplot(
    data=df[df['fraud_flag'] == 0],
    x='fraud_score',
    kde=True,
    stat="density",
    color='#2b7bba',
    label='Non-Fraudulent (0)',
    alpha=0.4,
    bins=30
)

sns.histplot(
    data=df[df['fraud_flag'] == 1],
    x='fraud_score',
    kde=True,
    stat="density",
    color='#d9534f',
    label='Fraudulent (1)',
    alpha=0.5,
    bins=30
)


plt.title('Fraud Risk Score Distribution by Actual Fraud Status', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Fraud Risk Score', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(axis='x', linestyle=':', alpha=0.6)
plt.legend(title='Ground Truth Label', fontsize=11)
plt.tight_layout()
plt.show()
```

* The area where the red and blue curves overlap represents the model's uncertainty zone.
* the blue curve should peak close to 0 and the red curve should peak close to the maximum possible score (e.g., 1)
* If the blue curve stretches far into the high-score territory, these are legitimate customers mistakenly flagged as high-risk


*  The comparison between Claim Status and Fraudulent flags shows that a portion of historical fraud claims successfully reached an Approved status rather than being Denied or trapped in Pending. These rows represent critical operational leakage points that should be isolated to retrain underwriting rules.
